# Data Demo Generator

To create a representative demo dataset while preserving the natural proportions of zodiac signs (12 fixed signs) and categories (general, love, career, welness consistent across days), we implemented a stratified random sampling approach by days. Since each day contains the same number of horoscopes, the proportions remain constant regardless of the sampling strategy. This simplifies generation: we randomly selected X days from the full corpus, extracting all associated horoscopes directly. This method ensures balanced coverage without bias toward specific signs or categories, yielding ~Y texts (12 signs × 3 categories × X days) for efficient testing of TF-IDF, MinHashing, and clustering pipelines. Sampling by days also maintains temporal realism, as horoscopes evolve daily while recycling common phrases.

# Libraries

```python
import pandas as pd
import numpy as np
```

# Data Input

```python
data = pd.read_csv('../data/horoscope_full.csv')
```

# Data Generator

```python
n_dates = len(data["date"].unique())
print(n_dates)

# 30 different integers between 0 and n_dates-1 (no replacement)
nums = np.random.choice(n_dates, size=30, replace=False)
nums
```

```python
data_sampled = data[data["date"].isin(data["date"].unique()[nums])]
```

```python
data_sampled.reset_index(drop=True, inplace=True)
```

```python
data_sampled.to_csv('../data/horoscope_demo.csv', index=False)
```
